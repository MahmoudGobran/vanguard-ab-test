```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Set display options for better viewing
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)
pd.set_option('display.float_format', '{:.4f}'.format) # Set float format for cleaner output

print("--- Starting Day 6: Further Analysis and Segmentation ---")

# --- Load the Cleaned and Merged DataFrame ---
processed_data_path = r'C:\Users\Mahmoud Gobran\00_Ironhack\Projects\vanguard-ab-test\data\processed'
input_file_name = 'df_full_ab_test_cleaned.csv'
input_full_path = os.path.join(processed_data_path, input_file_name)

try:
    df_full_ab_test = pd.read_csv(input_full_path)
    df_full_ab_test['date_time'] = pd.to_datetime(df_full_ab_test['date_time'])
    print(f"Successfully loaded df_full_ab_test from: {input_full_path}")

    # Re-apply the specific cleaning for demographic NaNs that was done in the previous notebook
    initial_ab_test_rows = len(df_full_ab_test)
    df_full_ab_test.dropna(subset=[
        'clnt_tenure_yr', 'clnt_tenure_mnth', 'clnt_age', 'gendr',
        'num_accts', 'bal', 'calls_6_mnth', 'logons_6_mnth'
    ], inplace=True)
    print(f"Removed {initial_ab_test_rows - len(df_full_ab_test)} rows with NaNs in demographic columns after reload.")
    print(f"df_full_ab_test now has {len(df_full_ab_test)} entries for analysis.")

except FileNotFoundError as e:
    print(f"Error loading file: {e}. Make sure '{input_file_name}' was saved correctly from the previous notebook.")
    exit()

# --- Consolidate to Visit-Level KPIs for Segmentation ---
# The KPIs are already calculated and merged in df_full_ab_test.
# We create a single visit-level DataFrame containing these KPIs and key demographics.
# For 'completed_process' and 'has_backward_movement', 'max' is appropriate as it returns True if any row in the visit has True.
# For 'time_spent_minutes', 'mean' is appropriate as it's already calculated per visit (and NaN for incomplete).

visit_level_kpis = df_full_ab_test.groupby([
    'client_id', 'visit_id', 'variation',
    'clnt_age', 'clnt_tenure_yr', 'gendr',
    'num_accts', 'bal', 'calls_6_mnth', 'logons_6_mnth'
]).agg(
    completed_process=('completed_process', 'max'),
    time_spent_minutes=('time_spent_minutes', 'mean'), # Mean will correctly pick the single value or NaN
    has_backward_movement=('has_backward_movement', 'max')
).reset_index()

print("\n--- Consolidated Visit-Level KPIs for Segmentation ---")
print(visit_level_kpis.head())
print(f"Visit-level KPIs DataFrame shape: {visit_level_kpis.shape}")


print("\n--- Demographic Segmentation ---")

# --- 1. Create Demographic Bins/Categories ---
# Apply binning directly to the visit_level_kpis DataFrame

# Age Bins
bins_age = [18, 25, 35, 45, 55, 65, 100]
labels_age = ['18-24', '25-34', '35-44', '45-54', '55-64', '65+']
visit_level_kpis['age_band'] = pd.cut(visit_level_kpis['clnt_age'], bins=bins_age, labels=labels_age, right=False)

# Tenure Bins
bins_tenure = [0, 1, 5, 10, 50] # 50 as an upper bound for practical purposes
labels_tenure = ['<1 Yr', '1-5 Yrs', '5-10 Yrs', '10+ Yrs']
visit_level_kpis['tenure_band'] = pd.cut(visit_level_kpis['clnt_tenure_yr'], bins=bins_tenure, labels=labels_tenure, right=False)

# Number of Accounts Bins
bins_num_accts = [0, 1, 3, 6, 100] # Assuming 100 is a practical upper bound
labels_num_accts = ['1 Account', '2-3 Accounts', '4-6 Accounts', '7+ Accounts']
visit_level_kpis['num_accts_band'] = pd.cut(visit_level_kpis['num_accts'], bins=bins_num_accts, labels=labels_num_accts, right=False)

# Logons 6 Months Bins
bins_logons = [0, 1, 5, 15, 30, 100] # Example tiers: 0, low, medium, high, very high
labels_logons = ['0', '1-4', '5-14', '15-29', '30+']
visit_level_kpis['logons_band'] = pd.cut(visit_level_kpis['logons_6_mnth'], bins=bins_logons, labels=labels_logons, right=False)

# Balance Bins - Use quartiles for bal_band for even distribution
# Calculate quantiles from the full df_full_ab_test for consistency if desired,
# or from visit_level_kpis if segment distribution is preferred for this notebook.
# Sticking with df_full_ab_test for consistency with previous notebook's calculation of balance quantiles.
bal_quantiles = df_full_ab_test['bal'].quantile([0, 0.25, 0.5, 0.75, 1]).values
labels_bal = [f'Q{i+1}' for i in range(len(bal_quantiles) - 1)]
visit_level_kpis['bal_band'] = pd.cut(visit_level_kpis['bal'], bins=bal_quantiles, labels=labels_bal, include_lowest=True, right=True)


# List of demographic segments to analyze
demographic_columns = ['gendr', 'age_band', 'tenure_band', 'num_accts_band', 'logons_band', 'bal_band']

# --- 2. Segmented KPI Calculation and Visualization ---

def plot_segmented_kpi(data_df, segment_col, kpi_name, value_col, is_rate=False, invert_bad=False, sort_by_diff=True):
    """
    Calculates and plots a KPI by variation across segments.

    Args:
        data_df (DataFrame): The DataFrame containing the KPI and demographic columns (visit_level_kpis).
        segment_col (str): The demographic column to segment by.
        kpi_name (str): Name of the KPI (for plot title/labels).
        value_col (str): The column containing the value to aggregate (e.g., 'completed_process', 'time_spent_minutes', 'has_backward_movement').
        is_rate (bool): True if the KPI is a rate (e.g., completion rate, error rate), False for averages.
        invert_bad (bool): True if a higher value for the KPI is bad (e.g., time spent, error rate).
        sort_by_diff (bool): True to sort segments by the difference between Test and Control.
    """
    if data_df.empty:
        print(f"Skipping plot for {kpi_name} by {segment_col}: Data is empty.")
        return

    # Ensure the segment_col is in the DataFrame and doesn't have NaNs for grouping
    plot_df = data_df.dropna(subset=[segment_col])
    if plot_df.empty:
        print(f"Skipping plot for {kpi_name} by {segment_col}: No valid data after dropping NaNs in segment column for {segment_col}.")
        return

    if is_rate:
        # Calculate rates more robustly
        # Count 'True' occurrences for the value_col (e.g., completed, has_backward_movement)
        # and total observations for each segment-variation group.
        # Then calculate the proportion and convert to percentage.
        agg_data = plot_df.groupby([segment_col, 'variation'])[value_col].agg(
            true_count='sum', # sum for boolean columns counts Trues
            total_count='size'
        ).unstack(fill_value=0)

        # Calculate rate, handle division by zero
        # Ensure 'Test' and 'Control' columns exist for both 'true_count' and 'total_count'
        for col_prefix in ['true_count', 'total_count']:
            if (col_prefix, 'Test') not in agg_data.columns:
                agg_data[(col_prefix, 'Test')] = 0
            if (col_prefix, 'Control') not in agg_data.columns:
                agg_data[(col_prefix, 'Control')] = 0

        # Calculate the percentages
        segmented_kpi = pd.DataFrame(index=agg_data.index)
        segmented_kpi['Test'] = (agg_data[('true_count', 'Test')] / agg_data[('total_count', 'Test')] * 100).fillna(0)
        segmented_kpi['Control'] = (agg_data[('true_count', 'Control')] / agg_data[('total_count', 'Control')] * 100).fillna(0)

        ylabel = f'{kpi_name} (%)'
    else:
        # For averages (time spent)
        segmented_kpi = plot_df.groupby([segment_col, 'variation'])[value_col].mean().unstack()
        ylabel = f'Average {kpi_name}'

        # Ensure 'Test' and 'Control' columns exist for averages, fill with NaN
        if 'Test' not in segmented_kpi.columns:
            segmented_kpi['Test'] = np.nan
        if 'Control' not in segmented_kpi.columns:
            segmented_kpi['Control'] = np.nan


    # Drop segments where both Test and Control are NaN (meaning no data for that segment in either group)
    segmented_kpi = segmented_kpi.dropna(subset=['Test', 'Control'], how='all')
    if segmented_kpi.empty:
        print(f"No complete data points for {kpi_name} by {segment_col} after handling missing variations and NaNs.")
        return

    # Calculate difference and sort
    segmented_kpi['difference'] = segmented_kpi['Test'] - segmented_kpi['Control']
    if sort_by_diff:
        # Sort based on the difference, considering 'invert_bad' for ascending/descending
        # For 'bad' KPIs, we want larger negative differences (Test < Control) at the top, so ascending is True
        # For 'good' KPIs, we want larger positive differences (Test > Control) at the top, so ascending is False
        segmented_kpi = segmented_kpi.sort_values(by='difference', ascending=invert_bad)

    # Plotting
    fig, ax = plt.subplots(figsize=(12, 6))
    segmented_kpi[['Control', 'Test']].plot(kind='bar', ax=ax, colormap='viridis') # Using colormap for distinct colors
    plt.title(f'{kpi_name} by {segment_col} and Variation', fontsize=16)
    plt.xlabel(segment_col, fontsize=12)
    plt.ylabel(ylabel, fontsize=12)
    plt.xticks(rotation=45, ha='right')
    plt.grid(axis='y', linestyle='--', alpha=0.7)
    plt.tight_layout()
    plt.show()

    print(f"\n--- {kpi_name} by {segment_col} ---")
    print(segmented_kpi[['Control', 'Test', 'difference']])


print("\n--- Segmented KPI Analysis ---")

# Loop through each demographic column and generate plots for all KPIs
for demo_col in demographic_columns:
    print(f"\nAnalyzing by: {demo_col}")

    # Completion Rate (using 'completed_process' from visit_level_kpis)
    plot_segmented_kpi(visit_level_kpis, demo_col, 'Completion Rate', 'completed_process', is_rate=True, invert_bad=False)

    # Average Time Spent (using 'time_spent_minutes' from visit_level_kpis)
    # Filter for only completed processes that have a non-null time_spent_minutes value
    time_spent_data = visit_level_kpis[visit_level_kpis['completed_process'] == True].dropna(subset=['time_spent_minutes'])
    if not time_spent_data.empty:
        plot_segmented_kpi(time_spent_data, demo_col, 'Time Spent (Minutes)', 'time_spent_minutes', is_rate=False, invert_bad=True)
    else:
        print(f"No completed visits with valid duration data available for segmentation by {demo_col} (Time Spent).")


    # Error Rate (using 'has_backward_movement' from visit_level_kpis)
    plot_segmented_kpi(visit_level_kpis, demo_col, 'Error Rate (Backward Movement)', 'has_backward_movement', is_rate=True, invert_bad=True)


print("\n--- Qualitative Discussion: Why Increased Time/Errors? ---")
print("Based on the overall results, the new design leads to a higher completion rate but also increases average time spent and the likelihood of backward movement (errors).")
print("\nPossible reasons for increased time spent and errors in the Test group:")
print("1.  **Increased Complexity:** The new design might have more steps, more information to process, or less intuitive navigation that requires users to spend more time or retrace steps.")
print("2.  **Unclear UI/UX Elements:** New buttons, labels, or layout might be confusing, leading users to pause, search, or click incorrectly (causing backward movements).")
print("3.  **Bugs or Performance Issues:** Although not directly visible in the data, new features can sometimes come with performance lags or hidden bugs that frustrate users and make them take longer or abandon/restart parts of the process.")
print("4.  **Information Overload:** If the new design introduces more choices or information upfront, users might feel overwhelmed, leading to slower decisions or errors.")
print("5.  **Lack of Clear Affordances:** If interactive elements don't clearly indicate their function (e.g., a button doesn't look like a button), users might struggle.")
print("6.  **New Features/Functionality:** If the 'Test' group introduced new features that require more interaction or decision-making (even if beneficial), this could naturally extend the time spent.")
print("\nTo truly understand the 'why,' qualitative data (user interviews, usability testing, heatmaps, session recordings) would be invaluable.")

print("\n--- Further Analysis and Segmentation Complete ---")
```
